Write a SQL query that returns information for each Region for the year 2018. The output should contain:
Region
UniqueVendors
TotalEntries
UniqueVendors should represent the number of unique vendors in each region.
TotalEntries should represent the total number of entries for each region.
The results should be ordered by TotalEntries in descending order.

```python
import sqlite3
import pandas as pd

# Read the csv to pandas and then to sql 
df = pd.read_csv("regional_entries.csv")

# Initialize an empty database 
conn = sqlite3.connect("regional_entries.sqlite")

# Create a table
query = """
    CREATE TABLE IF NOT EXISTS regional_entries (
        EntryID INT PRIMARY KEY,
        Region VARCHAR(20),
        Vendor VARCHAR(50),
        EntryDate DATE,
        Year INT
    )
"""

conn.executescript(query)
print("Table created successfully!")

# Insert data 
df.to_sql("regional_entries", con=conn, if_exists="replace", index=False)
print("Data inserted successfully!")


# Solution query 
query = """
    SELECT
        Region,
        COUNT(DISTINCT Vendor) AS UniqueVendors,
        COUNT(*) AS TotalEntries
    FROM regional_entries
    WHERE Year = 2018
    GROUP BY Region
    ORDER BY
        TotalEntries DESC
"""

pd.read_sql(query, conn)
```
