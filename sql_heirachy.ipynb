Your query should determine the company's heirachical structure based on the position and ReportsTo columns. The output should include the Position of the individual, their FirstName and LastName and the Position of the person they report to. Exclude entries where ReportsTo is null. The result should be ordered by Position of the person they report to and then by the individuals position in ascending order. The expected columns in the output are Position, FirstName, LastName and ReportsToPosition

```python
import sqlite3
import pandas as pd

df = pd.read_csv("company_hierarchy.csv")

# create a database 
conn = sqlite3.connect("company.db")

# create a table to store employee information 
create_table = """
    CREATE TABLE IF NOT EXISTS employees (
        employee_id INT PRIMARY KEY,
        Position VARCHAR(30),
        FirstName VARCHAR(20),
        LastName VARCHAR(20),
        ReportTo VARCHAR(50)
    )
"""
# Insert rows into the table 
df.to_sql("employees", con=conn, if_exists="replace", index=False)

# =================
# Answer Query
# =================

query = """
    SELECT
        e.Position,
        e.FirstName,
        e.LastName,
        m.Position AS ReportsToPosition
    FROM employees e
    JOIN employees m
        ON e.ReportsTo = CONCAT(m.FirstName, " ", m.LastName)
    WHERE e.ReportsTo IS NOT NULL
    ORDER BY
        m.Position ASC, e.Position ASC
"""

pd.read_sql(query, conn)
```
